## PR 7 Google Search

```python

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

driver = webdriver.Chrome()
driver.get("https://www.google.com")
driver.find_element(By.NAME, "q").send_keys("Selenium WebDriver", Keys.ENTER)
print("Page Title:", driver.title)
driver.quit()
  
```

## PR 8 Login Automation

```python

from selenium import webdriver
from selenium.webdriver.common.by import By

driver = webdriver.Chrome()
driver.get("https://the-internet.herokuapp.com/login")
driver.find_element(By.ID, "username").send_keys("tomsmith")
driver.find_element(By.ID, "password").send_keys("SuperSecretPassword!")
driver.find_element(By.CSS_SELECTOR, "button[type='submit']").click()
print("Login Message:", driver.find_element(By.ID, "flash").text)
driver.quit()
  
```

## Pr 9 Locators

```python

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

d = webdriver.Chrome()
d.get("https://the-internet.herokuapp.com/login")
d.find_element(By.ID, "username").send_keys("tomsmith")
d.find_element(By.XPATH, "//input[@id='password']").send_keys("SuperSecretPassword!")
d.find_element(By.CSS_SELECTOR, "button[type='submit']").click()
d.get("https://www.google.com")
d.find_element(By.NAME, "q").send_keys("Selenium WebDriver", Keys.ENTER)
d.quit()
  
```
